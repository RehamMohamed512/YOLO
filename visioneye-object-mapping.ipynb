# VisionEye Object Mapping Using YOLO

## Objective

The objective of this task is to detect objects in a video
and visualize their relationship with a specific observation point.

VisionEye draws lines between the selected observation point
and detected objects.

```python
import cv2
import os
import matplotlib.pyplot as plt

from ultralytics.solutions import VisionEye
from google.colab import files
```

```python
uploaded = files.upload()

video_path = list(uploaded.keys())[0]

print("Uploaded video:", video_path)
```

```python
cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    raise Exception("Could not open video!")

width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = cap.get(cv2.CAP_PROP_FPS)

print("Width:", width)
print("Height:", height)
print("FPS:", fps)

cap.release()
```

## Video Information

Video quality affects object detection accuracy.

High-quality videos usually provide clearer object details.

Frame width and height are required to create the output video.

FPS represents the number of frames displayed per second.

Some videos may produce poor detections because of:
- Low resolution.
- Poor lighting.
- Blurry objects.
- Object occlusion.

```python
vision_point = (width // 2, height // 4)

visioneye = solutions.VisionEye(
    model="yolo11n.pt",
    classes=[0, 2],
    vision_point=vision_point,
    show=False
)

print("VisionEye initialized successfully!")
print("Vision Point:", vision_point)
```

## VisionEye Initialization

VisionEye uses a fixed observation point to visualize
the relationship between detected objects and that point.

### Class Filtering

Class filtering means selecting specific object categories.

We selected:
- Person
- Car

Filtering helps reduce unnecessary detections
and makes the output easier to understand.

### Difference Between Detection and VisionEye

Normal object detection draws bounding boxes around objects.

VisionEye also draws visual connections between
the observation point and detected objects.
